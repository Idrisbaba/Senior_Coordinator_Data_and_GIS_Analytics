## Objective

This notebook reconciles GPS-derived campaign coverage with the official electronic tally (e-tally) records to evaluate agreement between independent data sources. While the cleaned GPS tracks provide evidence of where vaccination teams travelled, the e-tally represents reported vaccination activity. Comparing these datasets allows potential programme gaps, equipment failures and reporting inconsistencies to be identified.

The workflow also aggregates settlement-level results to ward and Local Government Area (LGA) level, classifies discrepancies, and identifies statistically significant spatial clusters of missed settlements. These outputs provide operational evidence to support campaign monitoring and post-campaign decision making.

This notebook addresses Requirement 4 and Requirement 5 of the technical assessment by:

- Computing settlement, ward and LGA coverage.
- Reconciling GPS-derived settlement visits with reported vaccination activity.
- Quantifying discrepancies between independent data sources.
- Classifying plausible causes of disagreement.
- Identifying spatial clusters of missed settlements using spatial statistics.
- Producing reproducible outputs for reporting and mapping.
